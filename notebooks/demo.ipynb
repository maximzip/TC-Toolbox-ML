# toolbox_ml — Demo

Notebook de demostración del paquete `toolbox_ml` usando un dataset real.

```python
import seaborn as sns
import pandas as pd
from toolbox_ml.eda.core import (
    describe_df,
    tipifica_variables,
    get_features_num_regression,
    plot_features_num_regression,
    get_features_cat_regression,
    plot_features_cat_regression,
    detect_outliers,
)
```

## 1. Carga del dataset

```python
df = sns.load_dataset("titanic")
df.head()
```

## 2. Dimensiones del dataset

```python
df.shape
```

## 3. Resumen del DataFrame con describe_df

```python
describe_df(df)
```

## 4. Tipificación de variables

```python
tipifica_variables(df, umbral_categoria=10, umbral_continua=30.0)
```

## 5. Variables numéricas relacionadas con fare

```python
get_features_num_regression(df, target_col="fare", umbral_corr=0.2, pvalue=0.05)
```

## 6. Gráficos de variables numéricas

```python
plot_features_num_regression(df, target_col="fare", umbral_corr=0.2, pvalue=0.05)
```

## 7. Variables categóricas relacionadas con fare

```python
get_features_cat_regression(df, target_col="fare", pvalue=0.05)
```

## 8. Gráficos de variables categóricas

```python
plot_features_cat_regression(df, target_col="fare", pvalue=0.05, with_individual_plot=False)
```

## 9. Detección de outliers

```python
outliers = detect_outliers(df)
outliers
```
